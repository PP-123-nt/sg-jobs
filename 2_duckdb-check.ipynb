```python
import duckdb

# 1. Connect to DuckDB and create sgjob.db database
jobcon = duckdb.connect('sgjob.db')

# 2. Create new SQL table job_raw directly from the CSV
jobcon.execute("""
    CREATE OR REPLACE TABLE job_raw AS 
    SELECT * FROM read_csv_auto('SGJobData.csv')
""")

print("✅ DuckDB table 'job_raw' created successfully!")
```

```python
# Verify the table exists inside DuckDB, job_raw is SQL table created (not panda df)
row_count = jobcon.execute("SELECT COUNT(*) FROM job_raw").fetchone()[0]
print(f"Total rows in DuckDB: {row_count:,}")

# Ask DuckDB for the first 5 rows, .df() convert job_raw SQL table to a temp pandas df for display purposes
display(jobcon.execute("SELECT * FROM job_raw LIMIT 5").df())
```

```python
# Notice the quotes around the SQL query
row_count = jobcon.execute("SELECT COUNT(*) FROM job_raw;").fetchone()[0]
print(f"Total rows: {row_count}")
```

```python
# pull the entire SQL table job_raw into a Pandas DataFrame job_rawdf
job_rawdf = jobcon.execute("SELECT * FROM job_raw").df()

# use job_rawdf to run panda EDA checks
print(job_rawdf.shape)                         # 1. How big is your sample?
job_rawdf.info()                               # 2. Columns and data types
print(job_rawdf['average_salary'].describe().apply(lambda x: f"{x:,.2f}"))  # 3. Typical and extreme values
```

```python
job_rawdf.describe().T.style.format("{:,.2f}")
# job_rawdf.describe().T -> same output without format
```

```python
summary_jobraw_df = jobcon.execute("SUMMARIZE job_raw").df()
display(summary_jobraw_df)
```

```python
# check selected columns in summary_jobraw_df for approx count (different from unique count)
cols_to_check = [
    'employmentTypes', 
    'metadata_isPostedOnBehalf', 
    'metadata_repostCount', 
    'positionLevels', 
    'salary_type', 
    'status_jobStatus'
]

# filter rows based on the list, and only keep the columns you want to see
summary_filtered = summary_jobraw_df[summary_jobraw_df['column_name'].isin(cols_to_check)][['column_name', 'approx_unique']]
display(summary_filtered)
```

```python
# col is the variable name
# thic code ask python to list the unique type for all the columns selcted in cols_to_check query
# missing/NA/NaN records will be included

for col in cols_to_check:
    print(f"\n--- Unique types for: {col} ---")
    
    unique_type_check = f"""
        SELECT {col}, COUNT(*) as count 
        FROM job_raw 
        GROUP BY {col} 
        ORDER BY count DESC
    """
    display(jobcon.execute(unique_type_check).df())
```

```python
# if only want to check a specific column unique type and count

jobstatus_check = """
    SELECT status_jobStatus, COUNT(*) as count
    FROM job_raw
    GROUP BY status_jobStatus
    ORDER BY count DESC
"""

# Execute the query in DuckDB and display the results
display(jobcon.execute(jobstatus_check).df())
```

```python
# jobid is unique, check if there is any duplicate record by jobid
# Note: In Pandas missing values -> None or NaN | SQL/DuckDB missing value -> NULL
jobid_check = """
    SELECT metadata_jobPostId, COUNT(*) as count
    FROM job_raw
    GROUP BY metadata_jobPostId
    HAVING COUNT(*) > 1
    ORDER BY count DESC
"""
duplicates = jobcon.execute(jobid_check).df()
display(duplicates)
```

```python
# check missing jobid records
missing_jobid = """
    SELECT * 
    FROM job_raw 
    WHERE metadata_jobPostId IS NULL
"""
no_jobid_df = jobcon.execute(missing_jobid).df()

print(f"Found {len(no_jobid_df)} rows with missing Job IDs.")
```

```python
# create job_df to exclude rows without salary values
job_df = job_rawdf.dropna(subset=['metadata_jobPostId']).copy()

print('before:', len(job_rawdf))
print('after :', len(job_df))
print('blank rows removed:', len(job_rawdf) - len(job_df))

salary_check = ['salary_minimum', 'salary_maximum', 'average_salary']
job_df[salary_check].quantile([0.01, 0.05, 0.5, 0.95, 0.99, 0.999, 1]).round(0)
```

```python
import pandas as pd

# group minimum salary into bins to check how many records in those outlier
salary_bins   = [-1, 0, 99, 499, 999, 2999, 5999, 9999, 19999, 29999, 49999, float('inf')]
salary_range = ['0', '1-99', '100-499', '500-999', '1k-3k', '3k-6k', '6k-10k',
          '10k-20k', '20k-30k', '30k-50k', '50k+']

min_salary = pd.cut(job_df['salary_minimum'], bins=salary_bins, labels=salary_range)
summary = min_salary.value_counts().sort_index().to_frame('min_salary_count')

# summary.min_salary_count: count of each bin | len(job_df): total rows w/o blank
summary['pct_of_total'] = (summary.min_salary_count / len(job_df) * 100).round(2)  
summary
```

```python
# min salary < $1K and > $30K records
low_salary_df  = job_df[job_df.salary_minimum < 1000].copy()
high_salary_df = job_df[job_df.salary_minimum > 30000].copy()

print('low min salary < $1K:', f'{len(low_salary_df):,}')
print('high min salary >$30K:', f'{len(high_salary_df):,}')
```

```python
# min salary < $1K records by employment types
low_emp = low_salary_df['employmentTypes'].value_counts().to_frame('below_1k')
low_emp['pct'] = (low_emp.below_1k / len(low_salary_df) * 100).round(2)
low_emp
```

```python
# min salary < $1K records, vs total excl missing value, vs same employment type
low_emp = low_salary_df['employmentTypes'].value_counts().to_frame('low_rows')
low_emp['pct_of_low'] = (low_emp.low_rows / len(low_salary_df) * 100).round(2)

# total rows per employment type across the whole cleaned dataset
total_emp = job_df['employmentTypes'].value_counts().rename('total_rows')

low_emp = low_emp.join(total_emp)
low_emp['pct_of_total_rows'] = (low_emp.total_rows / len(job_df) * 100).round(2)
low_emp['low_as_pct_of_type'] = (low_emp.low_rows / low_emp.total_rows * 100).round(2)

# a total row at the bottom
low_emp.loc['TOTAL'] = [low_salary_df.shape[0], 100.0, len(job_df), 100.0,
                        round(low_salary_df.shape[0] / len(job_df) * 100, 2)]
low_emp
```

```python
# min salary > $30K records
high_pos = high_salary_df['positionLevels'].value_counts().to_frame('rows')
high_pos['pct'] = (high_pos.rows / len(high_salary_df) * 100).round(2)
high_pos
```

```python
import pandas as pd

yr_exp   = [-1, 0, 2, 5, 10, 15, 20, 30, 50, float('inf')]
yr_exp_range = ['0', '1-2', '3-5', '6-10', '11-15', '16-20', '21-30', '31-50', '50+']

yr_exp_bin = pd.cut(job_df['minimumYearsExperience'], bins=yr_exp, labels=yr_exp_range)
yr_exp_summary = yr_exp_bin.value_counts().sort_index().to_frame('count_of_jobs')
yr_exp_summary['pct_of_total'] = (yr_exp_summary['count_of_jobs'] / len(job_df) * 100).round(2)
yr_exp_summary['cum_pct_of_total'] = yr_exp_summary['pct_of_total'].cumsum().round(2)
yr_exp_summary
```

```python
import pandas as pd

# 1. month of the original posting date, as a label like 2023-05
post_month = job_df['metadata_originalPostingDate'].dt.to_period('M').astype(str)

# 2. count and share of total for each month
month_summary = post_month.value_counts().sort_index().to_frame('posting_count')
month_summary['pct_of_total'] = (month_summary['posting_count'] / len(job_df) * 100).round(2)
month_summary['cum_pct_of_total'] = month_summary['pct_of_total'].cumsum().round(2)

# 3. TOTAL row at the bottom
total_row = pd.DataFrame(
    {'posting_count': [month_summary['posting_count'].sum()],
     'pct_of_total': [100.00],
     'cum_pct_of_total': [None]},
    index=['TOTAL'])
month_summary = pd.concat([month_summary, total_row])

month_summary  # 2022-10 to 2023-04 are probably a ramp-up as posting starts from 172, 2023-05 onwards posting are stabilize at the range of 60K - 84K
```

```python
import pandas as pd

level_order = ['Fresh/entry level', 'Non-executive', 'Junior Executive', 'Executive',
               'Senior Executive', 'Professional', 'Manager', 'Middle Management', 'Senior Management']

# only trust salaries in a realistic monthly range
valid_min = job_df['salary_minimum'].between(1000, 30000)
valid_max = job_df['salary_maximum'].between(1000, 50000)
job_df['min_salary_clean'] = job_df['salary_minimum'].where(valid_min)
job_df['max_salary_clean'] = job_df['salary_maximum'].where(valid_max)

level_summary = job_df.groupby('positionLevels').agg(
    posting_count     = ('metadata_jobPostId', 'size'),
    valid_min_rows    = ('min_salary_clean', 'count'),
    valid_max_rows    = ('max_salary_clean', 'count'),
    median_min_salary = ('min_salary_clean', 'median'),
    median_max_salary = ('max_salary_clean', 'median'),
    median_min_exp_yr = ('minimumYearsExperience', 'median'),
).reindex(level_order)

level_summary['pct_of_total'] = (level_summary['posting_count'] / len(job_df) * 100).round(2)

# TOTAL row at the bottom
total_row = pd.DataFrame({
    'posting_count':     [len(job_df)],
    'valid_min_rows':    [valid_min.sum()],
    'valid_max_rows':    [valid_max.sum()],
    'median_min_salary': [job_df['min_salary_clean'].median()],
    'median_max_salary': [job_df['max_salary_clean'].median()],
    'median_min_exp_yr': [job_df['minimumYearsExperience'].median()],
    'pct_of_total':      [100.00],
}, index=['TOTAL'])

level_summary = pd.concat([level_summary, total_row])
level_summary
```
