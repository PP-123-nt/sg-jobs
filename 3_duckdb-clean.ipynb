```python
import duckdb

# Connect to the existing database
jobcon = duckdb.connect('sgjob.db') 

# Create the clean table FROM the existing job_raw SQL table 
# exclude: occupationid column since all w/o value and rows w/o value (based on jobpostid)
jobcon.execute("""
    CREATE OR REPLACE TABLE job_clean AS 
    SELECT * EXCLUDE (occupationId)
    FROM job_raw
    WHERE metadata_jobPostId IS NOT NULL
""")

print("✅ DuckDB table 'job_clean' created successfully!")

# Verify the counts (should be exactly 1,044,597)
display(jobcon.execute("SELECT COUNT(*) as clean_count FROM job_clean").df())
```

```python
# Double check to ensure job_clean table records has no missing value in any records
# 1. Get the summary of the clean table
summary_jobclean_df = jobcon.execute("SUMMARIZE job_clean").df()

# 2. Filter to show ONLY columns that still have nulls (null_percentage > 0)
columns_with_nulls = summary_jobclean_df[summary_jobclean_df['null_percentage'] > 0]

# 3. Display the results
if len(columns_with_nulls) == 0:
    print("🎉 Amazing! The clean table has ZERO missing values.")
else:
    print("⚠️ The following columns still have missing values:")
    display(columns_with_nulls[['column_name', 'null_percentage']])
```
